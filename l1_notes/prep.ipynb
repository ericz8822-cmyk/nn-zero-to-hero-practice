This notebook covers some most basic concepts before starting to learn about micrograd and other things about neural network, which is just about derivative and partial derivatives. 

# Derivative

```python
# plot of f(x)
def f(x):
    return 3*x**2 - 4*x +5
xs = np.arange(-5,5,0.25)
ys = f(xs)
plt.plot(xs, ys)

# plot of f'(x)
h = 0.0000001
def f_prime(x):
    return (f(x+h)-f(x))/h
xs = np.arange(-5,5,0.25)
ys = f_prime(xs)
plt.plot(xs,ys)
```

# Partial derivative

```python
# a multivariable function
def g(a,b,c):
    return a * b + c
print(g(1,2,3))

# the three partial derivatives
def g_partial(var, a,b,c):
    match var:
        case "a":
            return (g(a+h,b,c)-g(a,b,c))/h
        case "b": 
            return (g(a,b+h,c)-g(a,b,c))/h
        case "c": 
            return (g(a,b,c+h)-g(a,b,c))/h
print(g_partial("a", 1,2,3))
```
